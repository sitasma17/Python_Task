# Exploring len() on Slices

Continue using lang = "Programming".

Slice lang to get "Program" and store it in a variable part1.

Print len(part1) and comment how it differs from len(lang).

```python
lang = 'Programming'

part1 = lang[0:7]
print(part1)

print(len(part1))
print(len(lang))

# len(part1) is 7 because 'Program' has 7 characters,
# while len(lang) is 11 because 'Programming' has 11 characters.
# Slicing creates a shorter new string, so part1 length is less.
```

# In-Place vs. Reassignment with String Methods

Let word = "example".

Call word.upper() but do not store it, then print word.

Now set word = word.upper(), then print word.

Comment on why the second print is different from the first.

```python
word = 'example'

word.upper()
print(word)

word = word.upper()
print(word)

#  first one just calls .upper() and throws away the result so word stays "example",
# but second one saves the uppercase result back into word so now word becomes "EXAMPLE"
```

# Arithmetic Operator Precedence

Define a = 5, b = 3, c = 2.

Print the result of the expression a + b * c.

Print the result of (a + b) * c.

In comments, explain how operator precedence affects the outcome.

```python
a=5
b=3
c=2

print(a+b*c)

print((a+b)*c)

# In first, * has higher precedence than +,so b * c = 6 is calculated first, then 5 + 6 = 11.
# In second, parentheses override precedence, so (a + b) = 8 is calculated first, then 8 * 2 = 16.
```
